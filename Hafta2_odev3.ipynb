```python
import cv2
resim = cv2.imread("gri.jpg", cv2.IMREAD_GRAYSCALE)
satir_sayisi, sütun_sayisi = resim.shape

orta_satir = satir_sayisi // 2
orta_sütun = sütun_sayisi // 2

for i in range(satir_sayisi):
 for j in range(sütun_sayisi):
  if i < orta_satir and j < orta_sütun:
   resim[i,j] = (resim[i,j]//64)* 64

  elif i < orta_satir and j >= orta_sütun:
            resim[i,j] = (resim[i, j]//16)* 16
  elif i >= orta_satir and j < orta_sütun:
            resim[i,j] = (resim[i, j] // 8)*8
else:
            resim[i,j] = resim[i,j]
cv2.imshow("4,16,32,256", resim)
cv2.waitKey(0)
```
