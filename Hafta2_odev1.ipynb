```python
import cv2
resim = cv2.imread("bw.jpeg", cv2.IMREAD_GRAYSCALE)
iki_katı = cv2.resize(resim, (0, 0), fx=2, fy=2)
dört_katı = cv2.resize(resim, (0, 0), fx=4, fy=4)
yarısı = cv2.resize(resim, (0, 0), fx=0.5, fy=0.5)
çeyrek = cv2.resize(resim, (0, 0), fx=0.25, fy=0.25)
cv2.imshow("İki Kat", iki_katı)
cv2.imshow("Dört Kat", dört_katı)
cv2.imshow("Yarısı", yarısı)
cv2.imshow("Çeyrek", çeyrek)
cv2.waitKey(0)
```
