```python
import cv2
resim = cv2.imread("gri.jpg", cv2.IMREAD_GRAYSCALE)
satir_sayisi, sütun_sayisi = resim.shape

resim128 = resim.copy()
for i in range(satir_sayisi):
 for j in range(sütun_sayisi):
  resim128[i,j] = (resim128[i, j]// 2)*2

resim64 = resim.copy()
for i in range(satir_sayisi):
 for j in range(sütun_sayisi):
  resim64[i,j] = (resim64[i, j]// 4)*4

resim4 = resim.copy()
for i in range(satir_sayisi):
    for j in range(sütun_sayisi):
        resim4[i,j] = (resim4[i, j]// 64)*64

cv2.imshow("128 deger", resim128)
cv2.imshow("64 deger", resim64)
cv2.imshow("4 deger", resim4)
cv2.waitKey(0)
```
