```python
import cv2
resim = cv2.imread("mavi.jpg", cv2.IMREAD_GRAYSCALE)
satir_sayisi, sütun_sayisi = resim.shape
histogram = [0] * 256

for i in range(satir_sayisi):
  for j in range(sütun_sayisi):
     parlaklik = resim[i, j]
     histogram[parlaklik]+= 1
for i in range(256):
    print(i, "Piksel sayısı:", histogram[i])
```
